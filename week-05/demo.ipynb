## 固定四类初始对象

目的：看清数据怎样组织，分别查看类型与数量。

list 按顺序放元素；dict 用键关联值；set 不重复、不提供位置索引。NumPy 本次只用一维数值数组。

```python
import numpy as np

xs = [1, 2, 2, -1]
record = {"a": 1, "b": 2, "c": 2, "d": -1}
unique = {1, 2, 2, -1}
a = np.array([1, 2, 2, -1])

print("list：", xs, type(xs), len(xs))
print("dict：", record, type(record), len(record))
print("set：", unique, type(unique), len(unique))
print("array：", a, type(a), len(a))
print("数组形状与数据类型：", a.shape, a.dtype)
```

## 取值与成员判断

目的：区分位置、键与成员判断。`in` 用在字典上时检查键，不是值。

```python
xs = [1, 2, 2, -1]
record = {"a": 1, "b": 2, "c": 2, "d": -1}
unique = {1, 2, 2, -1}
a = np.array([1, 2, 2, -1])

print("列表位置 0：", xs[0])
print("数组位置 0：", a[0])
print("字典键 b：", record["b"])
print("集合中有 2：", 2 in unique)
print("字典中有键 b：", "b" in record)
print("字典中有键 2：", 2 in record)
```

## 修改与增加

```python
a = np.array([1, 2, 2, -1])
a[0] = 10
grown = np.append(a, 5)
print("原数组：", a)
print("追加后的新数组：", grown)
print("两个数组的长度：", len(a), len(grown))
```

### 微练习 3｜函数／方法，修改／返回

先看下面代码的原列表和返回值。只把 `xs.append(5)` 改成 `xs.pop()`：目标是原列表剩下 `[1, 2, 2]`，返回值为 `-1`。运行核对两项，不只看其中一项。

```python
xs = [1, 2, 2, -1]
returned = xs.pop()
print("原列表现在是：", xs)
print("这次调用返回：", returned)
```

## 删除与取出

目的：核对删掉谁、原对象是否改变、返回什么。set 的 `pop()` 取任意元素，不预测具体取到哪个，也不比较显示顺序。

```python
xs = [1, 2, 2, -1]
removed = xs.pop()
print("list 取出：", removed, "剩下：", xs)

record = {"a": 1, "b": 2, "c": 2, "d": -1}
removed = record.pop("b")
print("dict 取出值：", removed, "剩下：", record)

unique = {1, 2, 2, -1}
returned = unique.discard(2)
print("discard 后还有 2：", 2 in unique, "返回：", returned)
removed = unique.pop()
print("set 取出：", removed)
print("取出的值已不在集合：", removed not in unique)

a = np.array([1, 2, 2, -1])
remaining = np.delete(a, 0)
print("原数组：", a, "删去首项的新数组：", remaining)
```

```python
a = np.array([1, 2, 2, -1])
grown = np.append(a, 0)
print("原数组：", a, len(a))
print("新数组：", grown, len(grown))
```

## 同时看原对象与返回值

- `len(xs)`：内置函数；`xs.append(5)`：列表对象的方法；`np.append(a, 5)`：NumPy 模块中的函数。
- `append` 方法改原列表、返回 `None`；`np.append` 不改原数组、返回新数组；`pop` 同时改原列表并返回元素。

## 遍历时，每次取出了什么？

目的：先看循环变量接收什么，再做计算。第二课时从下面的干净初态重新开始。

```python
xs = [1, 2, 2, -1]
record = {"a": 1, "b": 2, "c": 2, "d": -1}
unique = {1, 2, 2, -1}
a = np.array([1, 2, 2, -1])

print("遍历 list")
for item in xs:
    print(item)

print("遍历 dict")
for item in record:
    print(item)

print("遍历 set：顺序不作要求")
for item in unique:
    print(item)

print("遍历一维数组")
for item in a:
    print(item)
```

## 一项也可以是一对值

目的：用一个 tuple 和一个 zip 例子，看懂成对数据与拆包。

```python
pair = ("a", 1)
key, value = pair
print("tuple：", pair, type(pair))
print("拆开：", key, value)

print("zip 配对：")
for key, value in zip(["a", "b"], [1, 2]):
    print(key, value)

record = {"a": 1, "b": 2, "c": 2, "d": -1}
print("dict.items()：")
for key, value in record.items():
    print(key, value)
```

### 微练习 5｜拿到的是键，还是键值对？

只改下面循环体的 `print`，让每行先显示值、后显示键，例如第一行为 `1 a`。不要修改字典。

```python
record = {"a": 1, "b": 2, "c": 2, "d": -1}
for key, value in record.items():
    print(value, key)
```

## 逐项变换，收集结果

目的：对每项执行同一个计算，收集到新列表；原输入不修改。

```python
xs = [1, 2, 2, -1]
squares = []
for item in xs:
    squares.append(item ** 2)
print("原列表：", xs)
print("结果：", squares)
```

### 微练习 6｜只换一种处理方式

把下面的“平方”改成“乘以 10”，其他结构不变。目标结果为 `[10, 20, 20, -10]`，原 xs 仍为 `[1, 2, 2, -1]`。

```python
xs = [1, 2, 2, -1]
result = []
for item in xs:
    result.append(item * 10)
print("原列表：", xs)
print("结果：", result)
```

## 加入条件，只处理需要的元素

目的：先筛选原元素，再执行变换并收集结果。

```python
xs = [1, 2, 2, -1]
positive_squares = []
for item in xs:
    if item > 0:
        positive_squares.append(item ** 2)
print("原列表：", xs)
print("结果：", positive_squares)
```

### 微练习 7｜换一份带零的输入

先预测下面的结果，再只把筛选条件改成“保留所有非负数”。新目标结果为 `[1, 0, 4]`。比较 `> 0` 和 `>= 0`，不要改平方计算。

```python
xs = [1, 0, 2, -1]
result = []
for item in xs:
    if item >= 0:
        result.append(item ** 2)
print(result)
```

## 用推导式生成新结果

目的：将已理解的循环写成推导式；对照结果类型、重复值和保留规则。本组重新建立固定初态。

```python
xs = [1, 2, 2, -1]
record = {"a": 1, "b": 2, "c": 2, "d": -1}
a = np.array([1, 2, 2, -1])

new_list = [x ** 2 for x in xs if x > 0]
new_set = {x ** 2 for x in xs if x > 0}
new_dict = {k: v ** 2 for k, v in record.items() if v > 0}
new_array = np.array([x ** 2 for x in a if x > 0])

print("新 list：", new_list)
print("新 set：", new_set)
print("新 dict：", new_dict)
print("新 array：", new_array)
print("原输入仍是：", xs, record, a)
```

### 微练习 8｜换一个筛选条件

只改下面的 if 条件，保留原值至少为 2 的记录，再计算平方。目标为 `{"b": 4, "c": 4}`，原 record 不变。

```python
record = {"a": 1, "b": 2, "c": 2, "d": -1}
result = {k: v ** 2 for k, v in record.items() if v >= 2}
print("原字典：", record)
print("新字典：", result)
```

### 综合题 B｜应用一批修改，报告最终变化

- 输入：原字典、有序更新 tuple 列表、删除键列表。先依次更新，再删除；同一键最后一次更新生效，缺失删除忽略。
- 返回：新 `record` 字典与最终相比原记录发生变化的键集合 `changed`。所有输入保持。

### 综合题 C｜对齐两次测量，找出超阈值变化

- 输入：唯一编号列表、两次等长一维整数数组、非负阈值；同一位置属于同一编号，允许空输入。

- 只有变化量的绝对值 **大于** 阈值才异常；等于阈值正常。构造一个负向异常和一个恰好等于阈值的例子。
